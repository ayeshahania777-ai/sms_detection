```python
# ============================================================
# SPAM SMS CLASSIFICATION - COMPLETE MACHINE LEARNING PROJECT
# ============================================================
#
# Goal:
# Predict whether an SMS message is:
# 0 = HAM  (normal message)
# 1 = SPAM (unwanted/promotional message)
#
# Project Flow:
# 1. Data Cleaning
# 2. EDA (Exploratory Data Analysis)
# 3. Text Preprocessing
# 4. Text Vectorization
# 5. Model Building
# 6. Model Evaluation
# 7. Model Improvement
# 8. Ensemble Learning
# 9. Save the Model
# ============================================================


# ============================================================
# 1. IMPORT BASIC LIBRARIES AND LOAD DATA
# ============================================================

import numpy as np
import pandas as pd

# numpy:
# Used for numerical operations and arrays.

# pandas:
# Used to read, clean and manipulate tabular data.

# Read the SMS dataset
df = pd.read_csv('spam.csv')

# df now contains our complete dataset.


# Display 5 random rows
# WHY?
# To understand what our dataset looks like.
df.sample(5)


# Display number of rows and columns
# WHY?
# To know the size of our dataset.
print("Dataset Shape:", df.shape)


# ============================================================
# 2. DATA CLEANING
# ============================================================

# Display information about the dataset
# WHY?
# To check:
# - column names
# - number of rows
# - data types
# - missing values
df.info()


# ------------------------------------------------------------
# Remove unnecessary columns
# ------------------------------------------------------------

# The original dataset contains 3 unnecessary columns:
# Unnamed: 2, Unnamed: 3, Unnamed: 4
#
# These columns contain mostly empty/unwanted information.
#
# Therefore, we remove them.

df.drop(
    columns=['Unnamed: 2', 'Unnamed: 3', 'Unnamed: 4'],
    inplace=True
)


# ------------------------------------------------------------
# Rename columns
# ------------------------------------------------------------

# Original:
# v1 -> contains ham/spam
# v2 -> contains SMS text
#
# We rename them to meaningful names.

df.rename(
    columns={
        'v1': 'target',
        'v2': 'text'
    },
    inplace=True
)


# ------------------------------------------------------------
# Convert HAM/SPAM into numbers
# ------------------------------------------------------------

from sklearn.preprocessing import LabelEncoder

# Create a LabelEncoder object
encoder = LabelEncoder()

# Convert:
# ham  -> 0
# spam -> 1
#
# WHY?
# Machine Learning algorithms work with numerical values.

df['target'] = encoder.fit_transform(df['target'])


# Display first 5 rows
# WHY?
# To check whether our cleaning and encoding worked correctly.
df.head()


# ------------------------------------------------------------
# Check missing values
# ------------------------------------------------------------

# WHY?
# Missing values can cause problems during ML training.
#
# isnull() finds missing values.
# sum() counts them.

print("\nMissing Values:")
print(df.isnull().sum())


# ------------------------------------------------------------
# Check duplicate rows
# ------------------------------------------------------------

# WHY?
# Duplicate messages can make the model see the same example
# multiple times and can affect the learning process.

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())


# ------------------------------------------------------------
# Remove duplicate rows
# ------------------------------------------------------------

# keep='first':
# Keep the first occurrence and remove repeated copies.

df = df.drop_duplicates(keep='first')


# Check again
print("\nDuplicates after removal:")
print(df.duplicated().sum())


# Check final dataset size
print("\nShape after cleaning:")
print(df.shape)


# ============================================================
# 3. EDA - EXPLORATORY DATA ANALYSIS
# ============================================================

# EDA means:
# Exploring and understanding the dataset before training ML.
#
# We want to understand:
# - How many spam messages?
# - How many normal messages?
# - How long are the messages?
# - Are there useful patterns?


# ------------------------------------------------------------
# Check HAM and SPAM counts
# ------------------------------------------------------------

print("\nHAM / SPAM Count:")
print(df['target'].value_counts())


# ------------------------------------------------------------
# Visualize HAM vs SPAM
# ------------------------------------------------------------

import matplotlib.pyplot as plt

plt.figure(figsize=(6, 6))

plt.pie(
    df['target'].value_counts(),
    labels=['ham', 'spam'],
    autopct="%0.2f"
)

plt.title("HAM vs SPAM")
plt.show()

# WHY?
# A pie chart helps us visually understand the distribution.
#
# In this dataset, HAM messages are much more common than SPAM.
# This is called IMBALANCED DATA.


# ============================================================
# 4. NLTK SETUP
# ============================================================
#NLTK setup means setting up the Natural Language Toolkit, a Python library used to process human language/text.
import nltk

# If NLTK is not installed, run this separately once:
# !pip install nltk

# Download tokenizer data
nltk.download('punkt')


# ============================================================
# 5. CREATE ADDITIONAL FEATURES
# ============================================================

# ------------------------------------------------------------
# Number of characters
# ------------------------------------------------------------

# len() counts the number of characters in each SMS.
#
# WHY?
# Message length may contain useful information for ML.

df['num_characters'] = df['text'].apply(len)


# ------------------------------------------------------------
# Number of words
# ------------------------------------------------------------

# word_tokenize() breaks a sentence into tokens/words.
#
# len() then counts the number of tokens.

df['num_words'] = df['text'].apply(
    lambda x: len(nltk.word_tokenize(x))
)


# ------------------------------------------------------------
# Number of sentences
# ------------------------------------------------------------

# sent_tokenize() separates the text into sentences.
#
# len() counts the number of sentences.

df['num_sentences'] = df['text'].apply(
    lambda x: len(nltk.sent_tokenize(x))
)


# Display the new features
print("\nData with additional features:")
print(df.head())


# ------------------------------------------------------------
# Statistical summary
# ------------------------------------------------------------

# describe() gives:
# count, mean, standard deviation, minimum,
# 25%, 50%, 75%, maximum.

print("\nOverall Statistics:")
print(
    df[
        ['num_characters',
         'num_words',
         'num_sentences']
    ].describe()
)


# ------------------------------------------------------------
# Statistics for HAM messages
# ------------------------------------------------------------

print("\nHAM Statistics:")

print(
    df[df['target'] == 0][
        ['num_characters',
         'num_words',
         'num_sentences']
    ].describe()
)


# ------------------------------------------------------------
# Statistics for SPAM messages
# ------------------------------------------------------------

print("\nSPAM Statistics:")

print(
    df[df['target'] == 1][
        ['num_characters',
         'num_words',
         'num_sentences']
    ].describe()
)


# ============================================================
# 6. VISUALIZATION
# ============================================================

import seaborn as sns


# ------------------------------------------------------------
# Character distribution
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

# HAM message character distribution
sns.histplot(
    df[df['target'] == 0]['num_characters']
)

# SPAM message character distribution
sns.histplot(
    df[df['target'] == 1]['num_characters'],
    color='red'
)

plt.title("Character Distribution: HAM vs SPAM")
plt.show()


# ------------------------------------------------------------
# Word distribution
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

sns.histplot(
    df[df['target'] == 0]['num_words']
)

sns.histplot(
    df[df['target'] == 1]['num_words'],
    color='red'
)

plt.title("Word Distribution: HAM vs SPAM")
plt.show()


# ------------------------------------------------------------
# Pairplot
# ------------------------------------------------------------

# WHY?
# Shows relationships between numerical features.

sns.pairplot(df, hue='target')
plt.show()


# ------------------------------------------------------------
# Correlation heatmap
# ------------------------------------------------------------

# WHY?
# Correlation tells us how numerical features are related.

plt.figure(figsize=(8, 6))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True
)

plt.title("Correlation Heatmap")
plt.show()


# ============================================================
# 7. TEXT PREPROCESSING
# ============================================================

# Text preprocessing steps:
#
# 1. Convert text to lowercase
# 2. Tokenization
# 3. Remove special characters
# 4. Remove stop words
# 5. Remove punctuation
# 6. Stemming
#
# WHY?
# Raw text contains unnecessary variations and characters.
# Cleaning it makes the text easier for the ML algorithm to learn.


import string
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer


# Download stopwords
nltk.download('stopwords')

# Create PorterStemmer object
ps = PorterStemmer()


# ------------------------------------------------------------
# Create text preprocessing function
# ------------------------------------------------------------

def transform_text(text):

    # STEP 1: Convert text to lowercase
    # Example:
    # "FREE MONEY" -> "free money"

    text = text.lower()


    # STEP 2: Tokenization
    # Convert one sentence into individual tokens/words.

    text = nltk.word_tokenize(text)


    # STEP 3: Remove special characters
    #
    # isalnum() checks whether the token contains
    # letters or numbers.

    y = []

    for i in text:

        if i.isalnum():
            y.append(i)


    # Copy the cleaned list into text
    text = y[:]

    # Clear y so we can reuse it
    y.clear()


    # STEP 4: Remove stop words and punctuation
    #
    # Stop words are common words such as:
    # is, the, a, an, in, to, etc.
    #
    # They often don't provide much useful information
    # for this classification task.

    for i in text:

        if (
            i not in stopwords.words('english')
            and i not in string.punctuation
        ):
            y.append(i)


    # Copy cleaned words back to text
    text = y[:]

    # Clear y again
    y.clear()


    # STEP 5: Stemming
    #
    # Convert words to a common root-like form.
    #
    # Example:
    # loving -> love

    for i in text:

        y.append(ps.stem(i))


    # Convert list back into a single string
    return " ".join(y)


# ------------------------------------------------------------
# Test our preprocessing function
# ------------------------------------------------------------

sample_text = (
    "I'm gonna be home soon and i don't want to talk "
    "about this stuff anymore tonight"
)

print("\nOriginal Text:")
print(sample_text)

print("\nTransformed Text:")
print(transform_text(sample_text))


# ------------------------------------------------------------
# Apply preprocessing to every SMS
# ------------------------------------------------------------

# WHY?
# We need to clean every message in our dataset.

df['transformed_text'] = df['text'].apply(transform_text)


# Check the result
print("\nAfter Text Preprocessing:")
print(df[['text', 'transformed_text']].head())


# ============================================================
# 8. WORDCLOUD
# ============================================================

# WordCloud visually shows frequently occurring words.

from wordcloud import WordCloud


# Create WordCloud object
wc = WordCloud(
    width=500,
    height=500,
    min_font_size=10,
    background_color='white'
)


# ------------------------------------------------------------
# SPAM WORDCLOUD
# ------------------------------------------------------------

# Select only spam messages
# Combine their transformed text
# Generate WordCloud

spam_wc = wc.generate(
    df[df['target'] == 1]['transformed_text'].str.cat(sep=" ")
)

plt.figure(figsize=(15, 6))
plt.imshow(spam_wc)
plt.axis('off')
plt.title("Common Words in SPAM Messages")
plt.show()


# ------------------------------------------------------------
# HAM WORDCLOUD
# ------------------------------------------------------------

ham_wc = wc.generate(
    df[df['target'] == 0]['transformed_text'].str.cat(sep=" ")
)

plt.figure(figsize=(15, 6))
plt.imshow(ham_wc)
plt.axis('off')
plt.title("Common Words in HAM Messages")
plt.show()


# ============================================================
# 9. FIND MOST COMMON SPAM WORDS
# ============================================================

spam_corpus = []

# Select only spam messages
for msg in df[df['target'] == 1]['transformed_text'].tolist():

    # Split every message into individual words
    for word in msg.split():

        # Add every word to spam_corpus
        spam_corpus.append(word)


# Check total number of words
print("\nTotal words in SPAM corpus:")
print(len(spam_corpus))


# ------------------------------------------------------------
# Count most common spam words
# ------------------------------------------------------------

from collections import Counter

# Counter counts how many times each word appears.

spam_word_counts = Counter(spam_corpus).most_common(30)

# Convert to DataFrame for plotting
spam_word_df = pd.DataFrame(
    spam_word_counts,
    columns=['word', 'count']
)


plt.figure(figsize=(12, 6))

sns.barplot(
    data=spam_word_df,
    x='word',
    y='count'
)

plt.xticks(rotation='vertical')
plt.title("Top 30 SPAM Words")
plt.show()


# ============================================================
# 10. FIND MOST COMMON HAM WORDS
# ============================================================

ham_corpus = []

for msg in df[df['target'] == 0]['transformed_text'].tolist():

    for word in msg.split():

        ham_corpus.append(word)


print("\nTotal words in HAM corpus:")
print(len(ham_corpus))


ham_word_counts = Counter(ham_corpus).most_common(30)

ham_word_df = pd.DataFrame(
    ham_word_counts,
    columns=['word', 'count']
)


plt.figure(figsize=(12, 6))

sns.barplot(
    data=ham_word_df,
    x='word',
    y='count'
)

plt.xticks(rotation='vertical')
plt.title("Top 30 HAM Words")
plt.show()


# ============================================================
# 11. TEXT VECTORIZATION USING TF-IDF
# ============================================================

# Machine Learning models cannot directly understand text.
#
# Example:
# "free prize win"
#
# must be converted into numbers.
#
# This process is called VECTORIZATION.
#
# We use TF-IDF to convert the text into numerical features.


from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer


# CountVectorizer is created here as in the original code.
# It is not used later in this particular implementation.
cv = CountVectorizer()


# Create TF-IDF vectorizer
#
# max_features=3000 means:
# Use a maximum of 3000 text features.

tfidf = TfidfVectorizer(max_features=3000)


# Convert transformed text into numerical vectors
X = tfidf.fit_transform(
    df['transformed_text']
).toarray()


# Check shape
#
# Rows    = number of SMS messages
# Columns = number of text features

print("\nShape of X:")
print(X.shape)


# ============================================================
# 12. CREATE TARGET VARIABLE
# ============================================================

# X = Input/features
# y = Output/target

y = df['target'].values


# ============================================================
# 13. TRAIN-TEST SPLIT
# ============================================================

from sklearn.model_selection import train_test_split


# Split dataset:
#
# 80% -> training data
# 20% -> testing data
#
# Training data:
# Used by the model to learn.
#
# Testing data:
# Used to check how well the model performs
# on unseen messages.

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=2
)


print("\nTraining data shape:")
print(X_train.shape)

print("\nTesting data shape:")
print(X_test.shape)


# ============================================================
# 14. NAIVE BAYES MODELS
# ============================================================

from sklearn.naive_bayes import (
    GaussianNB,
    MultinomialNB,
    BernoulliNB
)

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    precision_score
)


# Create three different Naive Bayes models

gnb = GaussianNB()
mnb = MultinomialNB()
bnb = BernoulliNB()


# ------------------------------------------------------------
# Gaussian Naive Bayes
# ------------------------------------------------------------

# Train the model
gnb.fit(X_train, y_train)

# Predict test messages
y_pred1 = gnb.predict(X_test)

print("\n========== Gaussian NB ==========")

# Accuracy:
# Percentage of all predictions that are correct.
print("Accuracy:",
      accuracy_score(y_test, y_pred1))

# Confusion matrix:
# Shows correct and incorrect predictions.
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred1))

# Precision:
# Out of messages predicted as spam,
# how many were actually spam?
print("Precision:",
      precision_score(y_test, y_pred1))


# ------------------------------------------------------------
# Multinomial Naive Bayes
# ------------------------------------------------------------

mnb.fit(X_train, y_train)

y_pred2 = mnb.predict(X_test)

print("\n========== Multinomial NB ==========")

print("Accuracy:",
      accuracy_score(y_test, y_pred2))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred2))

print("Precision:",
      precision_score(y_test, y_pred2))


# ------------------------------------------------------------
# Bernoulli Naive Bayes
# ------------------------------------------------------------

bnb.fit(X_train, y_train)

y_pred3 = bnb.predict(X_test)

print("\n========== Bernoulli NB ==========")

print("Accuracy:",
      accuracy_score(y_test, y_pred3))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred3))

print("Precision:",
      precision_score(y_test, y_pred3))


# ============================================================
# 15. TRY MANY DIFFERENT ML ALGORITHMS
# ============================================================

# WHY?
# We don't know beforehand which algorithm will perform
# best for our dataset.
#
# Therefore, we train multiple algorithms and compare them.


from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.ensemble import (
    RandomForestClassifier,
    AdaBoostClassifier,
    BaggingClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier
)

from xgboost import XGBClassifier


# ------------------------------------------------------------
# Create ML models
# ------------------------------------------------------------

svc = SVC(
    kernel='sigmoid',
    gamma=1.0
)

knc = KNeighborsClassifier()

mnb = MultinomialNB()

dtc = DecisionTreeClassifier(
    max_depth=5
)

lrc = LogisticRegression(
    solver='liblinear',
    penalty='l1'
)

rfc = RandomForestClassifier(
    n_estimators=50,
    random_state=2
)

abc = AdaBoostClassifier(
    n_estimators=50,
    random_state=2
)

bc = BaggingClassifier(
    n_estimators=50,
    random_state=2
)

etc = ExtraTreesClassifier(
    n_estimators=50,
    random_state=2
)

gbdt = GradientBoostingClassifier(
    n_estimators=50,
    random_state=2
)

xgb = XGBClassifier(
    n_estimators=50,
    random_state=2
)


# ------------------------------------------------------------
# Store all models in a dictionary
# ------------------------------------------------------------

clfs = {

    'SVC': svc,
    'KN': knc,
    'NB': mnb,
    'DT': dtc,
    'LR': lrc,
    'RF': rfc,
    'AdaBoost': abc,
    'BgC': bc,
    'ETC': etc,
    'GBDT': gbdt,
    'xgb': xgb
}


# ============================================================
# 16. FUNCTION TO TRAIN AND EVALUATE A MODEL
# ============================================================

def train_classifier(
    clf,
    X_train,
    y_train,
    X_test,
    y_test
):

    # Train the model
    clf.fit(X_train, y_train)

    # Predict test data
    y_pred = clf.predict(X_test)

    # Calculate accuracy
    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    # Calculate precision
    precision = precision_score(
        y_test,
        y_pred
    )

    # Return both values
    return accuracy, precision


# ============================================================
# 17. TRAIN ALL CLASSIFIERS
# ============================================================

accuracy_scores = []
precision_scores = []


# Loop through every model
for name, clf in clfs.items():

    # Train model and get its scores
    current_accuracy, current_precision = train_classifier(
        clf,
        X_train,
        y_train,
        X_test,
        y_test
    )

    # Display results
    print("\nFor", name)

    print(
        "Accuracy -",
        current_accuracy
    )

    print(
        "Precision -",
        current_precision
    )

    # Store scores
    accuracy_scores.append(current_accuracy)
    precision_scores.append(current_precision)


# ============================================================
# 18. CREATE MODEL PERFORMANCE TABLE
# ============================================================

performance_df = pd.DataFrame({

    'Algorithm': clfs.keys(),

    'Accuracy': accuracy_scores,

    'Precision': precision_scores

}).sort_values(
    'Precision',
    ascending=False
)


print("\n========== MODEL PERFORMANCE ==========")

print(performance_df)


# ============================================================
# 19. VISUALIZE MODEL PERFORMANCE
# ============================================================

# Convert the DataFrame into a format suitable for plotting.

performance_df1 = pd.melt(
    performance_df,
    id_vars="Algorithm"
)


plt.figure(figsize=(12, 6))

sns.barplot(
    data=performance_df1,
    x='Algorithm',
    y='value',
    hue='variable'
)

plt.ylim(0.5, 1.0)

plt.xticks(rotation='vertical')

plt.title("Algorithm Performance")

plt.show()


# ============================================================
# 20. MODEL IMPROVEMENT
# ============================================================

# The original project experiments with changing:
#
# - TF-IDF max_features
# - scaling
# - adding number of characters
#
# The purpose is to see whether changing the input features
# improves model performance.
#
# These experiments can be performed separately and compared.


# ============================================================
# 21. VOTING CLASSIFIER
# ============================================================

# Ensemble learning:
#
# Instead of depending on one model,
# combine predictions from multiple models.
#
# Here we combine:
# 1. SVM
# 2. Multinomial Naive Bayes
# 3. Extra Trees


svc = SVC(
    kernel='sigmoid',
    gamma=1.0,
    probability=True
)

mnb = MultinomialNB()

etc = ExtraTreesClassifier(
    n_estimators=50,
    random_state=2
)


from sklearn.ensemble import VotingClassifier


# Create Voting Classifier
#
# voting='soft':
# Uses predicted probabilities from the models
# to make the final decision.

voting = VotingClassifier(

    estimators=[
        ('svm', svc),
        ('nb', mnb),
        ('et', etc)
    ],

    voting='soft'
)


# Train voting classifier
voting.fit(
    X_train,
    y_train
)


# Predict test data
y_pred = voting.predict(X_test)


# Evaluate
print("\n========== VOTING CLASSIFIER ==========")

print(
    "Accuracy:",
    accuracy_score(y_test, y_pred)
)

print(
    "Precision:",
    precision_score(y_test, y_pred)
)


# ============================================================
# 22. STACKING CLASSIFIER
# ============================================================

# Stacking is another ensemble technique.
#
# Several models make predictions.
# A final model learns how to combine those predictions.


estimators = [

    ('svm', svc),

    ('nb', mnb),

    ('et', etc)

]


# Final model which learns from the predictions
final_estimator = RandomForestClassifier()


from sklearn.ensemble import StackingClassifier


# Create stacking classifier
clf = StackingClassifier(

    estimators=estimators,

    final_estimator=final_estimator
)


# Train stacking classifier
clf.fit(
    X_train,
    y_train
)


# Predict
y_pred = clf.predict(X_test)


# Evaluate
print("\n========== STACKING CLASSIFIER ==========")

print(
    "Accuracy:",
    accuracy_score(y_test, y_pred)
)

print(
    "Precision:",
    precision_score(y_test, y_pred)
)


# ============================================================
# 23. SAVE THE VECTORIZER AND MODEL
# ============================================================

# WHY SAVE THEM?
#
# We don't want to train the model every time the application
# starts.
#
# We save the trained objects so that later our website/app
# can load them and make predictions.


import pickle


# Save TF-IDF vectorizer
#
# This is important because a new SMS must be converted
# into the SAME numerical format used during training.

pickle.dump(
    tfidf,
    open('vectorizer.pkl', 'wb')
)


# Save the trained ML model
#
# Here we save Multinomial Naive Bayes,
# following the original notebook.

pickle.dump(
    mnb,
    open('model.pkl', 'wb')
)


print("\n======================================")
print("PROJECT COMPLETED SUCCESSFULLY!")
print("======================================")

print("\nFiles created:")

print("1. vectorizer.pkl")
print("2. model.pkl")

print("\nThese files can later be used by a website")
print("or application to predict whether a new SMS")
print("is HAM or SPAM.")
```
